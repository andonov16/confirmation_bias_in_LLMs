# Evaluating the result for all models

```python
import pandas as pd
```

```python
def print_results_from_log_df(df: pd.DataFrame) -> None:
    dcpm_cols = df.filter(like='DCPMI')

    df['Max DCPMI Name'] = dcpm_cols.idxmax(axis=1)
    df['Max DCPMI Value'] = dcpm_cols.max(axis=1)

    for column in df['Max DCPMI Name'].unique():
        print('-'*100)
        print('Cards:', column.split()[0])
        print('Number of times picked:', (df['Max DCPMI Name'] == column).sum())
```

## Load the results

```python
model_names = ['gpt-4o-mini', "gemma3_4b", "gemma3_12b", "gemma3_27b", "llama3.1_8b"]

for model_name in model_names:

    print(f"Model: {model_name}")
    print("=" * 40)

    logs_dfs_dict = {
        'Control': pd.read_csv(f'../logs/{model_name}_control_wason_study_results.csv'),
        'Anchoring': pd.read_csv(f'../logs/{model_name}_anchoring_wason_study_results.csv'),
        'Framing': pd.read_csv(f'../logs/{model_name}_framing_wason_study_results.csv'),
        'Position': pd.read_csv(f'../logs/{model_name}_position_wason_study_results.csv'),
    }

    for experiment_name, logs_df in logs_dfs_dict.items():
        print('='*100)
        print('EXPERIMENT:', experiment_name)
        print_results_from_log_df(logs_df)
```
